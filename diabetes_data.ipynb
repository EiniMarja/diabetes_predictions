## 3. Tutkimusaineiston kokoaminen

```python

demo = pd.read_sas("P_DEMO.xpt")

# 2. Tallenna CSV-muotoon
#demo.to_csv("P_DEMO.csv", index=False)
```

```python
demo_cols = [
    "SEQN", #tunniste (toimii avaimena aineistojen yhdistämisessä)
    "RIDAGEYR", #Ikä
    "RIAGENDR", #Sukupuoli
    "INDFMPIR" # Tulot_indeksi
]

df = demo[demo_cols].copy()


```

```python
DIQ= pd.read_sas("P_DIQ.xpt")
```

```python
DIQ_cols= [
    "SEQN",
    "DIQ010", #diagnosoitu_diabetes DIQ -> arvot (1 = diagnosoitu diabetes, 2 = ei diagnosoitua diabetesta, 3 = prediabetes / borderline, 7/9 = ei tietoa / kieltäytynyt.
]    
DIQ= DIQ[DIQ_cols]
```

```python
#Valitaan selitettäväksi muuttujaksi DIQ010, joka kuvaa, onko henkilöllä diagnosoitu diabetes
print(DIQ["DIQ010"].value_counts(dropna=False)) #tarkistetaan arvojen jakauma ja mahdolliset puuttuvat arvot selitettavästä muuttujasta
print(DIQ.head())
```

```python
df= df.merge(DIQ, on="SEQN", how="left")
```

```python
ghb = pd.read_sas("P_GHB.xpt")
```

```python
ghb_cols = [
    "SEQN",
    "LBXGH", #Pitkaaikaissokeri 
]
ghb = ghb[ghb_cols]
```

```python
df= df.merge(ghb, on="SEQN", how="left")
```

```python
hdl = pd.read_sas("P_HDL.xpt")
```

```python
hdl_cols = [
    "SEQN",
    "LBDHDDSI" # HDL     
]
hdl= hdl[hdl_cols]
```

```python
df = df.merge(hdl, on="SEQN", how="left")
```

```python

hscpr= pd.read_sas("P_HSCRP.xpt")
```

```python
hscpr_cols = [
    "SEQN",
    "LBXHSCRP", #CRP     
]
hscpr=hscpr[hscpr_cols]
```

```python
df= df.merge(hscpr, on="SEQN", how="left") 
```

```python
ins= pd.read_sas("P_INS.xpt")
```

```python
ins_cols = [
    "SEQN",
    "LBXIN", # nsuliini
]
ins=ins[ins_cols] 
```

```python
df = df.merge(ins, on="SEQN", how="left")
```

```python
tchol= pd.read_sas("P_TCHOL.xpt")
```

```python
tchol_cols = [
    "SEQN",
    "LBDTCSI", # Kokonaiskolesteroli       
]
tchol=tchol[tchol_cols]
```

```python
df = df.merge(tchol, on="SEQN", how="left")
```

```python
trigly= pd.read_sas("P_TRIGLY.xpt")
```

```python
trigly_cols = [
    "SEQN",
    "LBDTRSI", # Triglyseridit
    "LBDLDLSI" # LDL      
]
trigly=trigly[trigly_cols]
```

```python
df= df.merge(trigly, on="SEQN", how="left")
```

```python
bmx= pd.read_sas("P_BMX.xpt")
```

```python
bmx_cols = [
    "SEQN",
    "BMXBMI", # BMI"
    "BMXWAIST" # Vyotaronymparys"       
]
bmx=bmx[bmx_cols]
```

```python
df = df.merge(bmx, on="SEQN", how="left")
```

```python
PAQ= pd.read_sas("P_PAQ.xpt")
```

```python
PAQ_cols = [
    "SEQN",
    "PAQ605", # Reipas_liikunta
    "PAQ635", # Kavely
    "PAD680" # Istumisaika
]
PAQ=PAQ[PAQ_cols]
```

```python
df= df.merge(PAQ,on="SEQN",how="left")
```

```python
df.columns
```

```python
df = df.rename(columns={
    "SEQN": "Tunniste", "RIDAGEYR": "Ika", "RIAGENDR": "Sukupuoli", "INDFMPIR": "Varallisuusindeksi", "DIQ010": "Diabetes", "LBXGH": "Pitkaaikaissokeri", "LBDHDDSI": "HDL","LBXHSCRP": "CRP", "LBXIN": "Insuliini", "LBDTCSI": "Kokonaiskolesteroli",
    "LBDTRSI": "Triglyseridit", "LBDLDLSI": "LDL", "BMXBMI": "BMI", "BMXWAIST": "Vyotaronymparys","PAQ605": "Reipas_liikunta(pv/vk)", "PAQ635": "Kavely(pv/vk)", "PAD680": "Istumisaika/paikallaanolo"
})
```
